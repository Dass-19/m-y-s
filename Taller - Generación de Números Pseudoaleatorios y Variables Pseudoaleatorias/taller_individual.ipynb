```python
# =========================================================
# EJERCICIO 3
# GENERADOR CONGRUENCIAL LINEAL (GCL)
# =========================================================

print("\n================================================")
print("EJERCICIO 3")
print("================================================")

"""
INSTRUCCIONES:
-----------------------------------------------------------
Complete la implementación del generador congruencial lineal.

FÓRMULA:
X(n+1) = (a * Xn + c) mod m

PREGUNTAS:
-----------------------------------------------------------
- ¿Qué hace el operador %?
- ¿Qué ocurre si cambia la semilla?
- ¿Qué ocurre si cambia el módulo?
- ¿Los números parecen uniformes?
"""


def gcl(seed, a, c, m, n):

    numeros = []

    x = seed

    for i in range(n):

        # TODO:
        # implemente:
        # x = (a*x + c) % m
        x = (a * x + c) % m


        u = x / m

        numeros.append(u)

    return numeros


resultado = gcl(
    seed=20,
    a=5,
    c=3,
    m=16,
    n=20
)

print("Secuencia generada:")
print(resultado)
```

Por ejemplo, `5 % 2` devolverá `1`, ya que 5 dividido por 2 da un cociente de 2 con un resto de 1. En el contexto de la generación de números aleatorios, el operador `%` se puede usar para limitar los números generados a un rango específico, como `random_number % 10` para obtener un número entre 0 y 9.

- ¿Qué representa lambda?

Lambda en una distribución exponencial representa la tasa de eventos por unidad de tiempo. Es el parámetro que controla la forma de la distribución y determina la probabilidad de que un evento ocurra en un intervalo de tiempo específico. Un valor más alto de lambda indica una mayor tasa de eventos, lo que significa que los eventos ocurren con mayor frecuencia, mientras que un valor más bajo de lambda indica una menor tasa de eventos.

- ¿Dónde se usa esta distribución?

La distribución exponcial se utiliza en una variedad de campos para modelar el tiempo entre eventos en un proceso de Poisson. Algunos ejemplos incluyen:
1. En la ingeniería, para modelar el tiempo entre fallos de un sistema o componente
2. En la biología, para modelar el tiempo entre eventos como la muerte de células o la llegada de pacientes a un hospital
3. En la economía, para modelar el tiempo entre transacciones o eventos financieros

- ¿Por qué aparece el logaritmo?

El logaritmo aparece en la generación de números aleatorios a partir de una distribución exponencial debido a la forma en que se transforma un número aleatorio uniforme en un número aleatorio con la distribución deseada. La función de distribución acumulativa (CDF) de la distribución exponencial es F(x) = 1 - e^(-lambda * x). Para generar un número aleatorio con esta distribución, se puede usar la inversa de la CDF, que implica tomar el logaritmo natural. Esto se hace para transformar un número aleatorio uniforme en un número que siga la distribución exponencial, lo que es esencial para simular eventos que ocurren a tasas constantes.

```python
# =========================================================
# EJERCICIO 10
# SIMULACIÓN FINAL
# =========================================================

print("\n================================================")
print("EJERCICIO 10")
print("================================================")

"""
TRABAJO FINAL
-----------------------------------------------------------

Diseñe una simulación sencilla usando variables
pseudoaleatorias.

OPCIONES:
-----------------------------------------------------------
- supermercado
- hospital
- tráfico
- videojuegos
- cajero automático
- llamadas telefónicas
- entregas
- otra aprobada por el docente

REQUISITOS:
-----------------------------------------------------------
1. Debe usar generación pseudoaleatoria.
2. Debe usar al menos una distribución.
3. Debe incluir gráficos.
4. Debe incluir interpretación.
5. Debe explicar supuestos y limitaciones.

ENTREGABLE:
-----------------------------------------------------------
- Código funcional
- Explicación del problema
- Resultados
- Conclusiones
"""

# TODO:
# implemente aquí su simulación final

# Simulacion: cajero automatico (ATM)
# - Llegadas: proceso de Poisson (inter-arrival exponencial)
# - Servicio: tiempo de uso del cajero ~ uniforme

random.seed(42)

# Parametros
duracion_horas = 8
duracion_min = duracion_horas * 60
lamda_llegadas = 1 / 4  # en promedio 1 cliente cada 4 minutos
servicio_min = 2
servicio_max = 6

# Generar tiempos de llegada (inter-arrival)
llegadas = []
tiempo = 0.0
while tiempo < duracion_min:
    u = random.random()
    inter = -(1 / lamda_llegadas) * math.log(1 - u)
    tiempo += inter
    if tiempo <= duracion_min:
        llegadas.append(tiempo)

# Simular un solo cajero (cola FIFO)
inicio_servicio = []
fin_servicio = []
esperas = []

fin_ultimo = 0.0
for t_llegada in llegadas:
    t_servicio = random.uniform(servicio_min, servicio_max)
    t_inicio = max(t_llegada, fin_ultimo)
    t_fin = t_inicio + t_servicio
    inicio_servicio.append(t_inicio)
    fin_servicio.append(t_fin)
    esperas.append(t_inicio - t_llegada)
    fin_ultimo = t_fin

# Estadisticas
n_clientes = len(llegadas)
prom_espera = np.mean(esperas) if esperas else 0
max_espera = np.max(esperas) if esperas else 0
prom_servicio = (servicio_min + servicio_max) / 2

print(f"Clientes simulados: {n_clientes}")
print(f"Espera promedio (min): {prom_espera:.2f}")
print(f"Espera maxima (min): {max_espera:.2f}")
print(f"Tiempo promedio de servicio (min): {prom_servicio:.2f}")

# Grafico 1: Histograma de tiempos de espera
plt.figure(figsize=(7, 4))
plt.hist(esperas, bins=15, edgecolor='black')
plt.title("Cajero automatico - Histograma de espera")
plt.xlabel("Espera (min)")
plt.ylabel("Frecuencia")
plt.tight_layout()
plt.show()

# Grafico 2: Evolucion de la cola en el tiempo
eventos = []
for t in llegadas:
    eventos.append((t, 1))
for t in fin_servicio:
    eventos.append((t, -1))
eventos.sort(key=lambda x: x[0])

tiempos = []
cola = []
q = 0
for t, delta in eventos:
    q = max(0, q + delta)
    tiempos.append(t)
    cola.append(q)

plt.figure(figsize=(7, 4))
plt.step(tiempos, cola, where='post')
plt.title("Cajero automatico - Tamaño de cola")
plt.xlabel("Tiempo (min)")
plt.ylabel("Clientes en cola")
plt.tight_layout()
plt.show()
```

**Explicacion del problema**

Se simula un cajero automatico con un solo servidor durante 8 horas. Los clientes llegan de forma aleatoria y se atienden en orden de llegada.

**Supuestos**

- Las llegadas siguen un proceso de Poisson (inter-arrival exponencial).
- El tiempo de servicio es uniforme entre 2 y 6 minutos.
- Un solo cajero, cola FIFO, sin abandono.

**Resultados e interpretacion**

- El histograma muestra la variabilidad de las esperas: si la llegada es intensa, aumenta la probabilidad de colas.
- La grafica de cola permite ver picos de demanda y momentos de baja carga.

**Limitaciones**

- No considera clientes que se van por exceso de espera.
- No modela multiples cajeros ni cambios de demanda por hora.
- Los parametros son fijos y pueden no reflejar la realidad local.

**Conclusiones**

La simulacion permite estimar esperas y evaluar si un solo cajero es suficiente para la tasa de llegada supuesta. Con mayores tasas, las esperas crecen rapidamente.
